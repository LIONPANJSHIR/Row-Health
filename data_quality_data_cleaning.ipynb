### Data Quality Assessment — Table `campaigns`

#### 1. Présentation générale des données

La table `campaigns` contient **57 enregistrements et 8 variables** : `campaign_id`, `campaign_category`, `campaign_type`, `cost`, `platform`, `impressions`, `clicks` et `days_run`.

Chaque enregistrement représente une campagne marketing unique, identifiée par `campaign_id`.

Les campagnes sont réparties en **12 catégories thématiques** :

- Health For All
- Family Coverage Plan
- Golden Years Security
- Affordable Plans
- Tailored Health Plans
- Compare Health Coverage
- Benefit Updates
- Preventive Care News
- Summer Wellness Tips
- #HealthyLiving
- #CoverageMatters
- #InsureYourHealth

Elles sont diffusées sur **quatre plateformes marketing** : TV, SEO, Email et Social Media.

#### 2. Évaluation de la qualité des données

**A. Complétude — Valeurs manquantes**

L'analyse révèle **3 valeurs manquantes dans la variable `clicks`**, soit environ 5,3 % des enregistrements.

Ces valeurs concernent exclusivement les campagnes de la catégorie *Family Coverage Plan*, diffusées respectivement sur TV, SEO et Social Media.

Les autres variables ne présentent aucune valeur manquante.

**B. Cohérence métier — Mesure des clics selon les plateformes**

Les campagnes diffusées à la télévision enregistrent des valeurs nulles ou manquantes pour la variable `clicks`.

Cette particularité s'explique par la nature du canal TV, qui ne permet pas de mesurer directement les interactions par clic, contrairement aux plateformes digitales.

**Conséquence analytique :** le taux de clics (*Click-Through Rate — CTR*) ne sera pas utilisé pour comparer la télévision aux plateformes digitales. Les campagnes TV seront principalement évaluées à partir des impressions, des dépenses marketing et des inscriptions attribuées.

**C. Validité — Valeurs fractionnaires**

L'analyse met également en évidence **8 observations présentant des valeurs fractionnaires dans la variable `clicks`**.

Ces valeurs sont incohérentes avec la nature habituelle de cette variable, qui représente un nombre d'interactions individuelles et devrait donc être exprimée en nombres entiers.

En revanche, les 57 observations de la variable `impressions` sont des nombres entiers.

**Décision de traitement :**

- Conserver les valeurs initiales dans la couche de données brutes (*Bronze*) afin de garantir leur traçabilité.
- Identifier et signaler les observations fractionnaires comme des anomalies de qualité.
- Envisager un arrondi à l'entier le plus proche dans la couche nettoyée (*Silver*), après vérification de son impact sur les indicateurs marketing.
- Conserver les valeurs manquantes comme telles, sans les remplacer automatiquement par zéro.

**D. Unicité — Identification des campagnes**

Les 57 enregistrements possèdent un identifiant `campaign_id` unique. Aucun doublon n'a été identifié sur cette variable.

Elle peut donc être utilisée comme **clé primaire** de la table.

#### 3. Synthèse et implications pour l'analyse business

Dans l'ensemble, la table `campaigns` présente une structure exploitable pour l'analyse des performances marketing. Toutefois, certaines anomalies doivent être prises en compte afin de garantir la fiabilité des indicateurs.

| Dimension | Résultat | Traitement retenu |
|---|---|---|
| Unicité | Aucun doublon sur `campaign_id` | Clé primaire valide |
| Complétude | 3 valeurs manquantes dans `clicks` | Conservation des valeurs manquantes |
| Validité | 8 valeurs fractionnaires dans `clicks` | Signalement et investigation |
| Cohérence métier | Clics TV non directement mesurables | CTR TV considéré comme non applicable |
| Intégrité des mesures | Impressions exprimées en entiers | Aucune correction nécessaire |

**Conclusion :**

Ces contrôles permettent d'identifier les limites du dataset avant le calcul des KPI marketing.

Une attention particulière sera portée à la comparabilité des plateformes et au traitement des clics manquants ou fractionnaires, afin d'éviter toute interprétation biaisée des performances des campagnes.

L'objectif est de disposer de données suffisamment fiables pour analyser l'efficacité des investissements marketing et formuler des recommandations pertinentes concernant l'allocation du budget.

### Data Quality Assessment — Table `claims`

La table `claims` contient **49 998 observations et 7 variables** : `customer_id`, `claim_id`, `claim_date`, `product_name`, `purchase_page_ref`, `claim_amount` et `covered_amount`.

Chaque observation correspond à une **demande de remboursement effectuée par un client**. Un même client peut soumettre plusieurs demandes, ce qui implique une relation *One-to-Many (1:N)* entre les tables `customers` et `claims`.

L'analyse de la qualité des données a permis d'identifier les éléments suivants :

- **Unicité :** les 49 998 valeurs de `claim_id` sont uniques, ce qui permet d'utiliser cette variable comme clé primaire.
- **Diversité des données :** la table contient 18 produits distincts et 17 valeurs distinctes non manquantes pour `purchase_page_ref`, avant standardisation.
- **Complétude :** la variable `purchase_page_ref` comporte 12 valeurs manquantes (`NULL`), ainsi que des modalités `0` et `unknown`. Ces trois cas seront harmonisés sous la valeur `unknown` afin de standardiser les données et de faciliter les analyses.
- **Cohérence financière :** les variables `claim_amount` et `covered_amount` présentent chacune 73 observations égales à zéro. Les vérifications confirment que ces deux montants sont simultanément nuls pour les mêmes observations.
- **Traitement des montants :** ces valeurs égales à zéro seront conservées, car elles peuvent correspondre à des demandes sans montant financier associé et ne constituent pas nécessairement des anomalies.

**Conclusion :** la table présente une structure globalement cohérente et exploitable. Les principales transformations porteront sur la standardisation de `purchase_page_ref`, tandis que les montants égaux à zéro seront conservés conformément aux règles de traitement définies.

Ces décisions permettront de préserver l'intégrité des observations et de préparer les données pour l'analyse des **volumes de demandes, des montants couverts et des comportements des clients après leur acquisition**.

# PHASE 2: HARMONIZATION & NETTOYAGES DES DONNÉES

```python

# Conversion des variables numériques
numeric_cols = ["cost", "impressions", "clicks", "days_run"]
compagne_v2 = compagne.copy()
for col in numeric_cols:
    compagne_v2[col] = pd.to_numeric(
        compagne_v2[col],
        errors="coerce"
    )

# Identification des clics manquants
compagne_v2["clicks_missing_flag"] = (
    compagne_v2["clicks"].isna()
)

# Identification des clics fractionnaires
compagne_v2["clicks_fractional_flag"] = (
    compagne_v2["clicks"].notna()
    & (compagne_v2["clicks"] % 1 != 0)
)

# Vérification clics > impressions
compagne_v2["clicks_gt_impressions_flag"] = (
    compagne_v2["clicks"] > compagne_v2["impressions"]
)

# Le CTR TV n'est pas comparable au CTR digital
compagne_v2["ctr_applicable_flag"] = (
    (compagne_v2["platform"] != "TV")
    & compagne_v2["clicks"].notna()
)
```

```python

# Conversion des dates américaines
customers = customer.copy()
customers["signup_date"] = pd.to_datetime(
    customers["signup_date"],
    format="%m/%d/%y",
    errors="coerce"
)

# Harmonisation des campagnes inconnues
customers["campaign_id"] = (
    customers["campaign_id"]
    .replace(["unknown", ""], pd.NA)
)

# Standardisation
customers["plan"] = customers["plan"].str.lower()
customers["state"] = customers["state"].str.upper()

# Signalement des clients sans attribution
customers["campaign_unattributed_flag"] = (
    customers["campaign_id"].isna()
)
```
